```python
# Python is dynamically typed.
# This means that we do not have to declare the data type of a variable.
# Python determines the type automatically when the program runs.

x = 10
print(x)
print(type(x))

# The same variable can later store a value of a different type.
x = "Hello"
print(x)
print(type(x))

# This works because Python determines the type of x at runtime.
```
