```python
# Task 1
# Python is dynamically typed.
# This means that we do not have to declare the data type of a variable.
# Python determines the type automatically when the program runs.

x = 10
print(x)
print(type(x))

# The same variable can later store a value of a different type.
x = "Hello"
print(x)
print(type(x))

y = "Hi"
print(y)
print(type(y))

z = "Whats up?"
print(z)
print(type(z))

# This works because Python determines the type of x at runtime.

#Task 2
# We can use the type() function to check what type a variable has.
# This can be useful when we want to know what kind of data we are working with.

name = "Linus"
age = 25
height = 1.80

print(type(name))    # <class 'str'>
print(type(age))     # <class 'int'>
print(type(height))  # <class 'float'>

# Knowing the type can help us avoid errors and understand
# what operations we can perform on a variable.
```
